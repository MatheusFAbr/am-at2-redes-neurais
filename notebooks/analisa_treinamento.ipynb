# iFood Campaign Intelligence
## Análise Exploratória e Treinamento de Rede Neural MLP

### 1. Contexto do Problema

Este projeto tem como objetivo desenvolver um modelo de
Machine Learning capaz de prever a resposta de clientes
a campanhas de marketing.

A solução utiliza dados históricos de comportamento,
perfil demográfico e consumo dos clientes para identificar
padrões associados à aceitação de campanhas.

O problema será tratado como uma tarefa de classificação
binária, utilizando uma Rede Neural Multicamadas (MLP)
implementada com Scikit-Learn.

```python
import pandas as pd
import numpy as np

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Bibliotecas importadas com sucesso!")
```

```python
caminhos = [
    Path("data/dataset.csv"),
    Path("../data/dataset.csv")
]

DATA_PATH = next(
    (caminho for caminho in caminhos if caminho.exists()),
    None
)

if DATA_PATH is None:
    raise FileNotFoundError("Arquivo dataset.csv não encontrado.")

df = pd.read_csv(
    DATA_PATH,
    sep=None,
    engine="python",
    encoding="utf-8-sig"
)

print("Dataset carregado com sucesso!")
print(f"Arquivo: {DATA_PATH}")
```

```python
print(f"Linhas: {df.shape[0]}")
print(f"Colunas: {df.shape[1]}")

display(df.head())
```

```python
df.info()
```

```python
valores_ausentes = df.isnull().sum()

display(
    valores_ausentes[valores_ausentes > 0]
    .sort_values(ascending=False)
)

print("Total de valores ausentes:", df.isnull().sum().sum())
```
